```python
def srgb_to_linear(rgb: np.ndarray) -> np.ndarray:
    a = 0.055
    threshold = 0.04045
    return np.where(
        rgb <= threshold,
        rgb / 12.92,
        ((rgb + a) / (1 + a)) ** 2.4,
    )
```

```python
def normalize01(x: np.ndarray, eps: float = 1e-12) -> np.ndarray:
    x = x.astype(np.float32, copy=False)
    x_min = float(x.min())
    x_max = float(x.max())
    denom = (x_max - x_min) if (x_max - x_min) > eps else eps
    return (x - x_min) / denom
```

```python
def log_density(hist: np.ndarray) -> np.ndarray:
    return np.log1p(hist.astype(np.float32))    
```

```python
def overlay_crosshair(img_rgb: np.ndarray, thickness: int = 1) -> np.ndarray:
    h, w, _ = img_rgb.shape
    out = img_rgb.copy()

    cy, cx = h // 2, w // 2
    out[max(0, cy - thickness):min(h, cy + thickness + 1), :, :] = 1.0
    out[:, max(0, cx - thickness):min(w, cx + thickness + 1), :] = 1.0
    return out
```
